```python
# Named Entity Recognition
#Named Entity Recognition is an NLP technique used to find and identify important names or entities in a text


# Person → Rahul Gandhi Place → India Organization → Google Date → 10 August 2026 Money → ₹5000 Company → Microsoft Time → 10 AM
#Example--Mishika visited Delhi and bought a laptop from Amazon
#Mishika → Person Delhi → Location Amazon → Organization
#*Applications of NER
#1 Search Engines Helps  understand names of people places companies and other important information
#2 Chatbots understand important information from user message
#3 News Analysis Can identify people companies places and organizations mentioned in news articles
#4 Resume Analysis Can identify names skills companies degrees and locations from resumes
#5 Information Extraction Helps extract useful information from large amounts of text
#6 Customer Support Can identify products order numbers locations and other important details from customer messages
#7 Healthcare  Can identify diseases medicines symptoms and other medical terms from medical text 
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
from nltk import word_tokenize, pos_tag, ne_chunk
text = "my name is mishika, i was born on 6 may 2006"
tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)
named_entities = ne_chunk(pos_tags)
print(named_entities)
```

```python
# parts of speech tagging - used to identify the grammatical role of each word in a sentence Noun → Person place thing ,Verb → Action or state,Adjective → Describes a noun ,Adverb → Describes a verb adjective or another adverb ,Pronoun → Replaces a noun, Preposition → Shows relationship between words
#Applications of POS Tagging
#Helps in understanding sentence structure
#Used in Named Entity Recognition
#Used in machine translation
#Helps in text preprocessing
#Used in chatbots and question answering
#Helps in identifying important words in a sentence
```

```python
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
#text="i am learning nlp"
text ="She sells seashells by the seashore and Peter Piper picked a peck of pickled peppers"
tokens= nltk.word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)
named_entities = nltk.ne_chunk(pos_tags)
print(named_entities)
```
