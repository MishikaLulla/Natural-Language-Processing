# 1 jaccard Dictionary Based Method

Jaccard Distance is a method used to measure how different two sets of words are
It can be used in word correction to compare a misspelled word with possible correct words

Formula
 Jaccard Similarity=∣A∪B∣/∣A∩B∣
Jaccard Distance=1−Jaccard Similarity

A = {a, b, c, d}
B = {a, b, c, e}
Common elements are
A ∩ B = {a, b, c} - 3

A ∪ B = {a, b, c, d, e} - 5 unique elements

Jaccard Similarity = 3 / 5 = 0.6
similarity is 0.6 or 60 percent
Jaccard Distance = 1 − 0.6 = 0.4
the distance is 0.4
Smaller distance = more similar
Larger distance = less similar

APPLICATIONS
Spelling Correction
It can compare a misspelled word with possible correct words and find the most similar one
Text Similarity
It can measure how similar two texts are by comparing their words
Document Similarity
It can find documents that contain similar words or information
Duplicate Detection
It can help identify duplicate or very similar documents
Search Engines
It can compare a search query with documents and find similar results
Recommendation Systems
It can compare users or items based on common features or interests

# 2 Edit Distance Method
It finds the correct word by checking how many changes are needed to convert the wrong word into a correct word
The fewer the changes the more likely the word is the correct one

Three Basic Operations-

1 Insertion
Adding a character
cat → cart
Add r

2 Deletion
Removing a character
cart → cat
Remove r

3 Substitution
Replacing one character with another
cat → cut
Replace a with u


-Applications of Edit Distance


Spelling correction – finds the closest correct word
Example: helo → hello
Search engines – finds results even when the user makes spelling mistakes
Autocomplete – suggests similar words when typing
Plagiarism detection – compares how similar two texts are
Duplicate detection – identifies similar or repeated text
DNA sequence analysis – compares biological sequences
Natural Language Processing – used for comparing and matching words or text
