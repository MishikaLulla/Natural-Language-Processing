```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("spaCy loaded successfully!")
```

```python
text="Lemmatization is the process of converting a word into its base or dictionary form called a lemma"
doc = nlp(text)
print(f"{'Token':<15}{'lemma':<15}{'POS':<10}{'Explanation'}")
print("..............................................................")

for token in doc:
    print(f"{token.text:<15}{token.lemma_:<15}{token.pos_:<10}{spacy.explain(token.pos_)}")
```
