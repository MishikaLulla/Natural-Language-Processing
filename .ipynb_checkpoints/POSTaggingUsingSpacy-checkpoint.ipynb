```python
# POS Tagging using spaCy
#(Part-of-Speech) Tagging means identifying the grammatical role of each word in a sentence, such as noun, verb, adjective, pronoun, etc
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("spaCy loaded successfully!")
```

```python
text = "Tagging means identifying the grammatical role of each word in a sentence, such as noun, verb, adjective, pronoun, etc"

doc = nlp(text)

print(f"{'Token':<15}{'POS':<10}{'DTag':<15}{'Explanation'}")
print("..............................................................")

for token in doc:
    print(f"{token.text:<15}{token.pos_:<10}{token.tag_:<15}{spacy.explain(token.tag_)}")
```
