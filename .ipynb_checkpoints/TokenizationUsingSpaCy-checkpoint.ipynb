# Text Cleaning and Preprocessing using SpaCy
SpaCy is a Python library used for Natural Language Processing (NLP). It provides tools to clean and preprocess text before further analysis.
Main steps
Convert text to lowercase
Makes words consistent
Example: Hello → hello

Tokenization
Splits text into individual words or tokens
Example: I love Python → I, love, Python

Remove stop words
Removes common words that may not add much meaning
Example: the, is, a, and

Remove punctuation

Removes symbols such as ., ,, !, ?

Remove unnecessary spaces

Cleans extra spaces from the text


# Tokenization using SpaCy
Tokenization is the process of breaking text into smaller units called tokens such as words, punctuation marks, and numbers

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("spaCy loaded successfully!")
```

```python
text= " i am learning about spacy in nlp which provides tools to clean and preprocess text before further analysis"
doc=nlp(text)
print("tokens:")
for tokens in doc:
    print(tokens.text)
```
