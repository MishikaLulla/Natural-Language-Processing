# Dependency Parsing using SpaCy
Dependency Parsing is a technique used in NLP to understand the relationship between words in a sentence
# Why is Dependency Parsing useful?
It helps NLP systems understand the structure and meaning of sentences
Information extraction
Question answering
Text analysis
Machine translation
Understanding relationships between words

token.text	-- The actual word
token.dep_	-- Dependency relationship of the word
token.head	-- The main word that the current word depends on
ROOT	-- Main word or central part of the sentence

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("spaCy loaded successfully!")
```

```python
text = "The cat eats fish."
doc = nlp(text)
print(f"{'Token':<15}{'head':<15}{'POS':<10}{'DTag':<15}{'Explanation'}")
print("..............................................................")

for token in doc:
    print(f"{token.text:<15}{token.head.text:<15}{token.pos_:<10}{token.tag_:<15}{spacy.explain(token.tag_)}")
```
