eg html leftouts or brand names and product names

# Problems with Rare Words
The model gets very little information about the word
It becomes difficult to understand its meaning
It can reduce the performance of the NLP model
New or uncommon words may be difficult for the model to process

# Method in NLTK
FreqDist()-Frequency Distribution
It is used in NLTK to count how many times each word or item occurs in a text

fdist.most_common()
Shows the most frequently occurring words

```python
import nltk
from nltk import FreqDist
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
text = "Natural language processing is used to understand human language. NLP helps computers process text and find useful information. Text preprocessing is an important part of NLP. Tokenization divides text into smaller tokens. Stemming reduces words to their root form. Lemmatization finds the meaningful base form of words. NLP is used in chatbots translation sentiment analysis and text classification. Machine learning helps computers learn patterns from text. Python libraries like NLTK make NLP tasks easier. NLTK can be used for tokenization stemming lemmatization and frequency analysis. NLP is useful in many applications and is an important field of artificial intelligence."
words = word_tokenize(text.lower())
fdist = FreqDist(words)
fil_word = [word for word in words if fdist[word] > 1]
fil_word
```
