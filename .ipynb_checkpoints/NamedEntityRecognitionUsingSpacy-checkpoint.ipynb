Named Entity Recognition is an NLP technique used to identify important names or entities in a text, such as people, places, organizations, dates, and money
NER helps computers understand who, where, when, and what is being mentioned in a text

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("spaCy loaded successfully!")
```

```python
text="Named Entity Recognition is an NLP technique used to identify important names or entities in a text, such as people, places, organizations, dates, and money"
doc = nlp(text)
print(f"{'entity text':<25}{'label':<15}{'Explanation'}")
print("..............................................................")

for ent in doc.ents:
    print(f"{ent.text:<25}{ent.label:<15}{spacy.explain(ent.label_)}")
```
