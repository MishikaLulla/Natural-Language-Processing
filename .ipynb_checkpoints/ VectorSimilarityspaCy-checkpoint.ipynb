* Vector similarity
is used to find out how similar two words, sentences or documents are in meaning
In spaCy, words and documents can be represented as vectors These vectors contain numerical values that represent the meaning or characteristics of the text

A vector is simply a list of numbers representing a word
Vector similarity tells us how close two vectors are

* Applications of Vector Similarity
Text classification
Search engines
Recommendation systems
Finding similar documents
Question answering
Plagiarism detection
Chatbots
Semantic search

* Semantic relationship
the relationship between the meanings of words or sentences
Dog →  Cat 
Dog and cat are semantically related because both are animals


* Vector Similarity Core Concept
Vector similarity measures how similar two pieces of text are by comparing their numerical vector representations

Steps
Suppose 
Dog
Cat
Car
Each word is converted into a vector a list of numbers representing its meaning
Dog → 0.8, 0.7, 0.2, ...
Cat → 0.7, 0.8, 0.3, ...
Car → 0.1, 0.2, 0.9, ...

now 
Dog ↔ Cat
   ↓
Vectors are close
   ↓
High similarity
Closer vectors → more similar meaning
Farther vectors → less similar meaning


* limitations of vector similarity
1-it may not understand the exact meaning of words in every context
2-similar words can sometimes have different meanings
3-different words can have similar meanings but may get a lower similarity score
the score depends on the quality of the trained model
4-it may not work well with new words slang or misspelled words
5-small or unsupported spaCy models may not have proper word vectors
6-similarity score tells how close the vectors are but does not always explain why they are similar
7-it can miss some relationships between words because meaning is more complex than just numbers

8-specialized jargon in legal or medical fields
vector similarity may not understand specialized terms used in legal or medical fields
words may have a specific meaning in that field which the model may not understand correctly
for example in medicine cardiac arrest and heart attack are related but they do not mean the same thing
in law terms like appeal, tort, or injunction have specific meanings that general language models may not capture accurately
a general model may therefore give an incorrect similarity score for specialized words

```python
import spacy

nlp = spacy.load("en_core_web_md")
```

```python
word1 = nlp("dog")
word2 = nlp("puppy")

print(word1.similarity(word2))
```

```python
import spacy
nlp = spacy.load("en_core_web_md")
sent1 = nlp("i enjoy painting")     #nlp() converts the sentence:
sent2 = nlp("i enjoy playing football")
similarity = sent1.similarity(sent2)

print(f"similarity score: {similarity:.2f}")  #to get accurate result 
```

```python
import spacy
nlp = spacy.load("en_core_web_md")
sent1 = nlp("i enjoy painting")     
sent2 = nlp("i love to dance")
similarity = sent1.similarity(sent2)

print(f"similarity score: {similarity:.2f}")
```
