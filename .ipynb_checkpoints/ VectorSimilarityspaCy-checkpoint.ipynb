* Vector similarity
is used to find out how similar two words, sentences or documents are in meaning
In spaCy, words and documents can be represented as vectors These vectors contain numerical values that represent the meaning or characteristics of the text

A vector is simply a list of numbers representing a word
Vector similarity tells us how close two vectors are

* Applications of Vector Similarity
Text classification
Search engines
Recommendation systems
Finding similar documents
Question answering
Plagiarism detection
Chatbots
Semantic search

* Semantic relationship
the relationship between the meanings of words or sentences
Dog →  Cat 
Dog and cat are semantically related because both are animals


* Vector Similarity Core Concept
Vector similarity measures how similar two pieces of text are by comparing their numerical vector representations

Steps
Suppose 
Dog
Cat
Car
Each word is converted into a vector a list of numbers representing its meaning
Dog → 0.8, 0.7, 0.2, ...
Cat → 0.7, 0.8, 0.3, ...
Car → 0.1, 0.2, 0.9, ...

now 
Dog ↔ Cat
   ↓
Vectors are close
   ↓
High similarity
Closer vectors → more similar meaning
Farther vectors → less similar meaning
