Vector similarity is used to find out how similar two words, sentences or documents are in meaning
In spaCy, words and documents can be represented as vectors These vectors contain numerical values that represent the meaning or characteristics of the text

A vector is simply a list of numbers representing a word
Vector similarity tells us how close two vectors are

-Applications of Vector Similarity
Text classification
Search engines
Recommendation systems
Finding similar documents
Question answering
Plagiarism detection
Chatbots
Semantic search
