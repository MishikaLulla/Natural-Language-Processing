4 Chatbots

5 Information Extraction

# pos word classes

2 Pronoun — replaces a noun
3 Verb — action or state
4 Adjective — describes a noun
5 Adverb — describes a verb adjective or another adverb
6 Preposition — shows relationship between words
7 Conjunction — joins words or sentences
8 Interjection — expresses emotion

# division of word classes in POS tagging
1 Open Class Words
These are words where new words can be added easily
Nouns ,Verbs ,Adjectives ,Adverbs

2 Closed Class Words
These are words where new words are rarely added
Pronouns,Prepositions,Conjunctions,Determiners,Auxiliary verbs

# Challenges in POS Tagging
POS tagging can be difficult because the same word can have different grammatical roles depending on the sentence

1 Ambiguous Words
A word can have different POS tags in different sentences
Example
I can swim
can → Verb
I opened the can
can → Noun

2 Context
The meaning and POS of a word can depend on the words around it
Example
I saw a flying bird
flying → Verb
I bought a flying machine
flying → Adjective

3 Unknown Words
New words slang names or technical terms may not be present in the training data
Example
ChatGPT
Instagram
new technical terms

4 Spelling Mistakes
Incorrect spelling can make it difficult for the model to identify the correct POS
Example
I am lerning Python

5 Multiple Meanings
Some words can have completely different meanings
Example
The bat is an animal
bat → Noun
He used a bat
bat → Noun
The correct meaning depends on the context

6 Different Languages
POS tagging becomes more challenging when dealing with different languages because grammar and sentence structures can be different

7 Informal Language
Social media and chat messages often contain slang abbreviations emojis and incomplete sentences
Example
lol ,brb ,gonna, clock it 🤏


# POS Tagging Approaches
three approaches

1 Rule Based Approach
In this approach POS tags are assigned using a set of predefined grammar rules
The rules are created based on the structure and grammar of the language

Example
The boy is running
The → Determiner
boy → Noun
is → Verb
running → Verb

Advantages
Easy to understand
Uses grammar rules
Works well when rules are clearly defined

Disadvantage
Requires many rules
Difficult to handle exceptions
Rules need to be created manually

2 Statistical Approach
The statistical approach uses probability and statistical methods to find the most likely POS tag for a word
It learns from a tagged training dataset

Example
If a word commonly appears as a noun in a particular context the model can give it a higher probability of being a noun

---- Machine Learning Approach
Machine learning based POS tagging learns patterns from training data instead of depending only on manually written rules
The model looks at
The word
Surrounding words
Context
Previous POS tags
and predicts the most suitable tag

Advantages
Can learn from data
Handles different contexts better
Does not require every rule to be written manually

Disadvantage
Requires a good training dataset
Accuracy depends on the quality of the training data

3 Transformation Based Approach
The Transformation Based Approach combines ideas from rule based and statistical approaches
It first assigns initial POS tags and then applies a series of rules or transformations to correct incorrect tags

Example
Initial Tag
      ↓
Check Context
      ↓
Apply Transformation Rule
      ↓
Correct POS Tag

For example a word may initially be tagged as a noun but a transformation rule can change it to a verb if the surrounding words indicate that it is being used as a verb

Advantages
Combines rules and learning
Can correct tagging errors
Uses contextual information


Rule Based → Uses predefined rules
Statistical or Machine Learning → Learns from data
Transformation Based → Starts with tags and improves them using transformations

# Steps Involved in Machine Learning Approach for POS Tagging
1 Collect Training Data
Collect a large amount of text where each word already has its correct POS tag
Example
I → Pronoun
love → Verb
Python → Noun

2 Preprocess the Data
Clean and prepare the text so it can be used by the model

3 Feature Extraction
Extract useful information from the text

Features can include
Current word
Previous word
Next word
Previous POS tag
Word suffix or prefix
Context of the word

4 Train the Model
Give the training data and features to a machine learning algorithm
The model learns the relationship between words and their POS tags

5 Test the Model
Give the trained model new text that it has not seen before

6 Predict POS Tags
The model predicts the most suitable POS tag for each word

Example
Mishika is learning Python
Mishika → Noun
is → Verb
learning → Verb
Python → Noun

7 Evaluate the Model
Compare the predicted tags with the correct tags and measure the performance of the model
